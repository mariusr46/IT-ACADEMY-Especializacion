### 1. Calculadora de l'índex de massa corporal

```python
def calculadora_imc(peso, altura):
    """
    Calcula el imc. El peso debe estar en kg y la altura en metros.
    """
    return round(peso / altura ** 2, 2)

def clasificar_imc(imc):
    """
    Clasifica el valor del imc en una categoría
    """
    if imc < 18.5:
        return "Peso bajo"
    elif imc < 25:
        return "Peso normal"
    elif imc < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"
    
def obtener_datos():
    """
    Pedimos y normalizamos los datos
    """
    try:
        peso = float(input("Introduce tu peso en kilos"))
        altura = float(input("Introduce tu peso en metros"))

    except ValueError:
        raise ValueError("El formato de los datos introducidos no es válido")

    if peso <= 0 or altura <= 0:
        raise ValueError("El peso y la altura deben ser superiores a 0")

    if peso > 300 or altura > 2.5:
        raise ValueError("El peso y la altura deben ser razonables")

    return peso, altura


def ejecucion_imc():
    """
    coordinamos las distintas funciones en una ejecución final
    """

    try:

        peso, altura = obtener_datos()
        resultado_imc = calculadora_imc(peso, altura)
        categoria = clasificar_imc(resultado_imc)

        print(f"Según tu peso de {peso} kg y tu altura de {altura} metros, tu imc es {resultado_imc} y por lo tanto tu estado corporal es de {categoria}")
    
    except ValueError as e:
        print("Error: ", e)
```

### 2. Conversor de temperaturas

```python
def convertir_temperatura(temperatura, tipo):

    conversiones = {}

    if tipo == "C":

        conversiones["Fahrenheit"] = round((temperatura * 9/5) +32, 2)
        conversiones["Kelvin"] = round(temperatura + 273.15, 2)

    elif tipo == "F":

        celsius = [temperatura - 32] * 5/9

        conversiones["Celsius"] = round(celsius, 2)
        conversiones["Kelvin"] = round(celsius + 273.15, 2)

    elif tipo == "K":

        celsius = [temperatura - 32] * 5/9

        conversiones["Celsius"] = round(celsius, 2)
        conversiones["Fahrenheit"] = round((temperatura * 9/5) + 32, 2)

    return conversiones


def pedir_temperatura():

    temperatura = float(input("Introduce la temperatura: "))

    return temperatura


def pedir_tipo():

    tipo = input("Introduce el tipo de medición: ")
    
    tipo = tipo.upper()

    return tipo


def validar_tipo(tipo):

    return tipo in ["C", "F", "K"]


def ejecutar_programa_temperaturas():

    try:

        temperatura = pedir_temperatura()
        tipo = pedir_tipo()

        if not validar_tipo(tipo):

            print("Error: Escala no válida. ")

            return


        resultado = convertir_temperatura(temperatura, tipo)

        print(f"La conversión de la temperatura {temperatura} de la escala {'Celsius' if tipo == 'C' else 'Kelvin' if tipo == 'K' else 'Fahrenheit'} es: ")
        print(resultado)

    except:

        print("Error: la temperatura debe ser introducida en formato numérico")

        
```

### 3. Contador de palabras

```python
def limpiar_texto(texto):

    texto = texto.lower()

    for puntuacion in string.punctuation:
        texto = texto.replace(puntuacion, "")

    return texto

def obtener_palabras(texto):

    return texto.split()

def contar_palabras(palabras):

    contador = {}

    for palabra in palabras:
        
        if palabra not in contador:

            contador[palabra] = 1

        else:

            contador[palabra] += 1

    return contador

def calcular_media(palabras):

    total_letras = 0

    for palabra in palabras:

        total_letras += len(palabra)

    if len(palabras) > 0:

        return round(total_letras / len(palabras), 2)

    else:
        return 0


def procesar_texto(texto):

    texto_limpio = limpiar_texto(texto)

    palabras = obtener_palabras(texto_limpio)

    frecuencia = contar_palabras(palabras)

    media = calcular_media(palabras)

    return frecuencia, media


def pedir_texto():

    texto = input("Introduce el texto a analizar: ")

    texto = texto.strip()

    return texto

def ejecutar_contador():

    try:

        texto_introducido = pedir_texto()

        if texto_introducido == "":

            print("Error: no has introducido ningún texto para poder analizar")

            return

        frecuencia, media = procesar_texto(texto_introducido)

        print(f"La frecuencia por palabras es {frecuencia} y la media de longitud de palabras es {media}.")

    except:

        print("Ha sucedido un error")
```

### 4. Diccionario invertido

```python
def invertir_diccionario(diccionario):
    invertido = {}
    repetidos = {}
    existen_duplicados = False

    for llave, valor in diccionario.items():
        if valor in invertido:

            existen_duplicados = True

            if not isinstance(invertido[valor], list):
                invertido[valor] = [invertido[valor]]
            invertido[valor].append(llave)
            repetidos[valor] = invertido[valor]
        else:
            invertido[valor] = llave

    if existen_duplicados:
        print(f"Existen valores repetidos y son los siguientes: {repetidos}")

    return invertido

    
diccionario = {'a': 1, 'b': 2, 'c': 3}
diccionario_invertido = invertir_diccionario(diccionario)
print(f"Diccionario final: {diccionario_invertido}")

diccionario = {'x': 'apple', 'y': 'banana', 'z': 'banana'}
diccionario_invertido = invertir_diccionario(diccionario)
print(f"Diccionario final: {diccionario_invertido}")
```

### 5. Contador y función de ordenar de palabras extraidas de texto

```python
def leer_archivo(ruta):
    with open(ruta, 'r', encoding='utf-8') as archivo:

        texto = archivo.read()

    return texto


def limpiar_texto(texto):

    texto = texto.lower()

    for puntuacion in string.punctuation:
        texto = texto.replace(puntuacion, "")

    return texto


def obtener_palabras(texto):

    return texto.split()


def contar_y_ordenar_palabras(palabras):

    diccionario = {}

    for palabra in sorted(palabras):

        letra_inicial = palabra[0]

        if letra_inicial not in diccionario:

            diccionario[letra_inicial] = {}

        if palabra in diccionario[letra_inicial]:

            diccionario[letra_inicial][palabra] += 1

        else:

            diccionario[letra_inicial][palabra] = 1

    return diccionario

    
def procesar_documento(nombre_documento):

    texto = leer_archivo(nombre_documento)

    texto_procesado = limpiar_texto(texto)

    palabras = obtener_palabras(texto_procesado)

    resultado = contar_y_ordenar_palabras(palabras)

    return resultado
```

### 6. Conversión de tipos de datos

```python
def conv_a_float(lista):

    convertidos = []
    no_convertidos = []

    for elemento in lista:
        if isinstance(elemento, (list, tuple)):

            for caracter in elemento:

                try:

                    numero = float(caracter)
                    convertidos.append(numero)

                except:
                    no_convertidos.append(caracter)

            
        else:

            try:
                
                numero = float(elemento)
                convertidos.append(numero)

            except:

                no_convertidos.append(elemento)

    return convertidos, no_convertidos

lista = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', 1, 1.4, 'not-a-number'), [1, 2, '3', '3.4']]

convertidos, no_convertidos = conv_a_float(lista)

print(f"Partiendo de la lista: {lista} \n"
        f"Los elementos convertidos son: {convertidos}\n"
        f"Y los elementos no convertidos son: {no_convertidos}")
```

### 7. Generador de contraseñas

```python
def contrasenya_inicial(mayusculas, minusculas, numeros, signos):

    caracteres = ""
    obligatorios = []

    if mayusculas:
        caracteres += string.ascii_uppercase
        obligatorios.append(np.random.choice(list(string.ascii_uppercase)))
    if minusculas:
        caracteres += string.ascii_lowercase
        obligatorios.append(np.random.choice(list(string.ascii_lowercase)))
    if numeros:
        caracteres += string.digits
        obligatorios.append(np.random.choice(list(string.digits)))
    if signos:
        caracteres += string.punctuation
        obligatorios.append(np.random.choice(list(string.punctuation)))

    return caracteres, obligatorios

def crear_contrasenya(longitud, mayusculas=True, minusculas=True, numeros=True, signos=True):

    caracteres, contrasenya = contrasenya_inicial(mayusculas, minusculas, numeros, signos)

    if len(caracteres) == 0:
        return "Error: debes activar al menos un tipo de caracter."
    if longitud < len(contrasenya):
        return f"Error: la longitud debe ser de {len(contrasenya)} caracteres o mas."

    lista_caracteres = list(caracteres)

    while len(contrasenya) < longitud:
        contrasenya.append(np.random.choice(lista_caracteres))

    np.random.shuffle(contrasenya)

    contrasenya_definitiva = "".join(contrasenya)

    pyperclip.copy(contrasenya_definitiva)

    return contrasenya_definitiva

nueva_contrasenya = crear_contrasenya(33, False, False, True, True)

print(f"Tu contraseña generada es: {nueva_contrasenya}")
```

### 8. Procesamiento de datos

```python
import pandas as pd
import re
from pathlib import Path


PATRON_RESULTADO = re.compile(r"^(\d+)\s*-\s*(\d+)$")


def leer_archivo(nombre_archivo):
    """
    Lee el archivo de texto y devuelve una lista con las líneas no vacías.

    Cada línea debería representar un partido con esta estructura:

    Equipo local    resultado    Equipo visitante

    Ejemplo:
    FC Barcelona    4-4    Badalona
    """

    ruta = Path(nombre_archivo)

    if not ruta.exists():
        raise FileNotFoundError(f"No se ha encontrado el archivo: {nombre_archivo}")

    with ruta.open("r", encoding="utf-8") as archivo:
        lineas = [linea.strip() for linea in archivo if linea.strip()]

    return lineas


def separar_linea_partido(linea):
    """
    Recibe una línea del archivo y extrae:

    - equipo local
    - equipo visitante
    - goles del equipo local
    - goles del equipo visitante

    Devuelve una tupla:

    (equipo_local, equipo_visitante, goles_local, goles_visitante)
    """

    partes_tabulador = linea.split("\t")

    if len(partes_tabulador) == 3:
        equipo_local = partes_tabulador[0].strip()
        resultado = partes_tabulador[1].strip()
        equipo_visitante = partes_tabulador[2].strip()

    else:
        partes = linea.split()

        indice_resultado = None

        for i, parte in enumerate(partes):
            if PATRON_RESULTADO.match(parte):
                indice_resultado = i
                break

        if indice_resultado is None:
            raise ValueError(f"No se ha podido encontrar un resultado válido en la línea: {linea}")

        equipo_local = " ".join(partes[:indice_resultado]).strip()
        resultado = partes[indice_resultado].strip()
        equipo_visitante = " ".join(partes[indice_resultado + 1:]).strip()

    if not equipo_local or not equipo_visitante:
        raise ValueError(f"Falta el nombre de algún equipo en la línea: {linea}")

    coincidencia = PATRON_RESULTADO.match(resultado)

    if coincidencia is None:
        raise ValueError(f"El resultado no tiene un formato válido en la línea: {linea}")

    goles_local = int(coincidencia.group(1))
    goles_visitante = int(coincidencia.group(2))

    return equipo_local, equipo_visitante, goles_local, goles_visitante


def crear_estadisticas_equipo():
    """
    Crea la estructura inicial de estadísticas para un equipo.

    Cada equipo empieza con todo a cero.
    """

    return {
        "partidos_ganados": 0,
        "partidos_empatados": 0,
        "partidos_perdidos": 0,
        "goles": 0,
        "goles_encajados": 0,
        "puntos": 0
    }


def actualizar_estadisticas(estadisticas, equipo, goles_marcados, goles_encajados):
    """
    Actualiza las estadísticas de un equipo concreto.

    Parámetros:
    - estadisticas: diccionario general con todos los equipos.
    - equipo: nombre del equipo que se va a actualizar.
    - goles_marcados: goles que ha marcado ese equipo en el partido.
    - goles_encajados: goles que ha recibido ese equipo en el partido.
    """

    if equipo not in estadisticas:
        estadisticas[equipo] = crear_estadisticas_equipo()

    estadisticas[equipo]["goles"] += goles_marcados
    estadisticas[equipo]["goles_encajados"] += goles_encajados

    if goles_marcados > goles_encajados:
        estadisticas[equipo]["partidos_ganados"] += 1
        estadisticas[equipo]["puntos"] += 3

    elif goles_marcados == goles_encajados:
        estadisticas[equipo]["partidos_empatados"] += 1
        estadisticas[equipo]["puntos"] += 1

    else:
        estadisticas[equipo]["partidos_perdidos"] += 1


def procesar_partidos(lineas):
    """
    Procesa todas las líneas del archivo.

    Devuelve:
    - estadisticas: diccionario con los datos acumulados de cada equipo.
    - lineas_invalidas: lista con las líneas que no se han podido procesar.
    """

    estadisticas = {}
    lineas_invalidas = []

    for numero_linea, linea in enumerate(lineas, start=1):

        try:
            equipo_local, equipo_visitante, goles_local, goles_visitante = separar_linea_partido(linea)

        except ValueError as error:
            lineas_invalidas.append((numero_linea, linea, str(error)))
            continue

        actualizar_estadisticas(
            estadisticas,
            equipo_local,
            goles_local,
            goles_visitante
        )

        actualizar_estadisticas(
            estadisticas,
            equipo_visitante,
            goles_visitante,
            goles_local
        )

    return estadisticas, lineas_invalidas


def construir_clasificacion(estadisticas):
    """
    Convierte el diccionario de estadísticas en un DataFrame de pandas.

    La clasificación se ordena por:
    1. puntos
    2. diferencia de goles
    3. goles marcados
    """

    df = pd.DataFrame.from_dict(
        estadisticas,
        orient="index"
    )

    df.index.name = "equipo"

    df["diferencia_goles"] = df["goles"] - df["goles_encajados"]

    df = df.sort_values(
        by=["puntos", "diferencia_goles", "goles"],
        ascending=[False, False, False]
    )

    return df


def extraer_resultados(df):
    """
    Extrae los resultados pedidos por el enunciado:

    - número total de goles por equipo
    - equipo más goleador
    - equipo más goleado
    """

    goles_por_equipo = df["goles"].to_dict()

    equipo_mas_goleador = df["goles"].idxmax()

    equipo_mas_goleado = df["goles_encajados"].idxmax()

    return goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado


def mostrar_resultados(df, goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado, lineas_invalidas):
    """
    Muestra por pantalla los resultados finales.
    """

    print("CLASIFICACIÓN GLOBAL:")
    print(df.to_string())

    print("\nNÚMERO TOTAL DE GOLES POR EQUIPO:")
    print(goles_por_equipo)

    print(f"\nEquipo más goleador: {equipo_mas_goleador}")
    print(f"Equipo más goleado: {equipo_mas_goleado}")

    if lineas_invalidas:
        print("\nLÍNEAS NO PROCESADAS:")

        for numero_linea, linea, motivo in lineas_invalidas:
            print(f"Línea {numero_linea}: {linea} -> {motivo}")


def main(nombre_archivo):
    """
    Función principal del programa.

    Coordina todo el proceso:
    1. Lee el archivo.
    2. Procesa los partidos.
    3. Construye la clasificación.
    4. Extrae los resultados pedidos.
    5. Muestra los resultados.
    6. Devuelve los datos por si se quieren reutilizar.
    """

    lineas = leer_archivo(nombre_archivo)

    estadisticas, lineas_invalidas = procesar_partidos(lineas)

    df = construir_clasificacion(estadisticas)

    goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado = extraer_resultados(df)

    mostrar_resultados(
        df,
        goles_por_equipo,
        equipo_mas_goleador,
        equipo_mas_goleado,
        lineas_invalidas
    )

    return df, goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado, lineas_invalidas


main("historic partits.txt");
```
