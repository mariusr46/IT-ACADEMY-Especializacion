```python
import numpy as np
import pandas as pd
import string
import pyperclip
import re
from pathlib import Path
```

```python
### 1 CALCULADORA ###

def recibir_datos():

    try: 
        peso = float(input("Introduce tu peso en kg"))
        altura = float(input("Introduce tu altura en kg"))

    except ValueError:
        raise ValueError("Peso y altura deben ser numericos")
    
    if peso <= 0 or peso > 500:
        raise ValueError(f"{peso} kilogramos no son razonables")
    
    if altura <= 0 or altura > 500:
        raise ValueError(f"{altura} metros no son razonables")

    return peso, altura
    
def calcular_imc(peso, altura):
    return round(peso / (altura ** 2), 2)

def categorizar_imc(imc):
    if imc < 18.5:
        return "Peso bajo"
    elif imc < 25:
        return "Peso normal"
    elif imc < 30:
        return "Sobrepeso"
    else:
        return "Obesidad"
    
def printear_resultado(peso, altura, imc, categoria):

    print(f"Según tu peso de {peso} kg y tu altura de {altura} metros, su imc es de {imc} y su categoria es: {categoria}")

def ejecutar_programa_imc():
    
    try:
    
        peso, altura = recibir_datos()
        imc = calcular_imc(peso, altura)
        categoria = categorizar_imc(imc)

        printear_resultado (peso, altura, imc, categoria)

    except ValueError as e:
        print(e)

    return peso, altura, imc, categoria

```

```python
ejecutar_programa_imc()
```

```python
### CONVERSOR DE TEMPERATURAS ###

def obtener_datos_limpios_temp():

    try:

        temperatura = float(input("Introduce la temperatura: "))
        medida = input("Introduce la medida (C, F o K)").upper()

    except ValueError:
        raise ValueError("La temperatura debe ser introducida en formato numérico")
    
    return temperatura, medida
    

def validar_datos_recibidos(temperatura, medida):
    if medida.upper() not in ["C", "F", "K"]:
        raise ValueError(f"La medida '{medida}' no está dentro de las aceptadas ('C', 'F', 'K')")
    
    if medida == "C" and temperatura < -273:
        raise ValueError(f"{temperatura} grados {medida} no son válidos")
    elif medida == "F" and temperatura < -500:
        raise ValueError(f"{temperatura} grados {medida} no son válidos")
    else:
        if temperatura < 0:
            raise ValueError(f"{temperatura} grados {medida} no son válidos")
    

def crear_conversor(temperatura, medida):

    conversiones = {}

    if medida == "C":
        conversiones["Fahrenheit"] = round(temperatura * 9/5 + 32, 2)
        conversiones["Kelvin"] = round(temperatura + 273.15, 2)

    elif medida == "F":
        celsius = round((temperatura - 32) * 5/9, 2)
        conversiones["Celsius"] =round(celsius, 2)
        conversiones["Kelvin"] = round(celsius + 273.15, 2)

    else:
        celsius = temperatura - 273.15
        conversiones["Celsius"] = round(celsius, 2)
        conversiones["Fahrenheit"] = round((celsius * 9/5) + 32, 2)

    return conversiones

def printear_resultado(temperatura, medida, conversiones):
    print(f"En la medida {medida} tenemos {temperatura} grados. El resto de conversiones son: ")

    for clave, valor in conversiones.items():
        print(f"{clave} -> {valor}")

def ejecutar_programa_temperaturas():

    try:

        temperatura, medida = obtener_datos_limpios_temp()
        validar_datos_recibidos(temperatura, medida)
        conversiones = crear_conversor(temperatura, medida)
        printear_resultado(temperatura, medida, conversiones)

        return conversiones
    
    except ValueError as e:
        print(e)
```

```python
### CONTADOR PALABRAS ###

def pedir_y_limpiar_texto():
    texto = input("Introduce tu texto: ")
    
    for puntuacion in string.punctuation:
        texto = texto.replace(puntuacion, "").lower()

    return texto

def obtener_palabras(texto):
    return texto.split()

def contar_palabras(palabras):

    contador = {}

    for palabra in palabras:
        if palabra in contador:
            contador[palabra] += 1
        else:
            contador[palabra] = 1

    return contador

def contar_media_palabra(palabras):
    letras = 0

    for palabra in palabras:
        letras += len(palabra)

    return round(letras / len(palabras), 2)

def ejecutar_programa_contador_palabras():

    texto = pedir_y_limpiar_texto()
    palabras = obtener_palabras(texto)
    contador = contar_palabras(palabras)
    media = contar_media_palabra(palabras)

    print(f"La frecuencia por palabras es \n{contador}")
    print(f"La media de longitud de las palabras es de {media} letras.")
        
```

```python
ejecutar_programa_contador_palabras()
```

```python
### DICCIONARIO INVERTIDO ###

def printear_diccionarios(invertido, repetido, existen_duplicados):

    print(f"El diccionario invertido es: \n{invertido}")

    if existen_duplicados:
        print(f"El diccionario de repetidos es:\n{repetido}")

    else:
        print("No existen duplicados")


def invertir_diccionario(diccionario):
    invertido = {}
    repetido = {}

    existen_duplicados = False

    for llave, valor in diccionario.items():
        if valor in invertido:

            existen_duplicados = True

            if not isinstance(valor, list):
                invertido[valor] = [invertido[valor]]
            invertido[valor].append(llave)
            repetido[valor] = invertido[valor]

        else:
            invertido[valor] = llave

    printear_diccionarios(invertido, repetido, existen_duplicados)

    return invertido, repetido
```

```python
### CONTADOR Y FUNCIÓN DE ORDENAR PALABRAS EXTRAÍDAS DE TEXTO ###

def leer_archivo_limpiar_texto(ruta):
    with open(ruta, 'r', encoding='utf-8') as archivo:
        texto = archivo.read()

    for puntuacion in string.punctuation:
        texto = texto.replace(puntuacion, "").strip().lower()

    return texto

def obtener_palabras(texto):
    return texto.split()

def ordenar_por_letra(palabras):

    diccionario = {}

    for palabra in sorted(palabras):
        letra = palabra[0]

        if letra not in diccionario:
            diccionario[letra] = {}

        if palabra in diccionario[letra]:
            diccionario[letra][palabra] += 1

        else:
            diccionario[letra][palabra] = 1

    return diccionario

def procesar_document(nombre_documento):
    texto = leer_archivo_limpiar_texto(nombre_documento)
    palabras = obtener_palabras(texto)
    diccionario = ordenar_por_letra(palabras)

    print(f"El conteo de palabras es \n{diccionario}")

    return diccionario
```

```python
### GENERADOR DE CONTRASEÑAS ###

def definir_caracteres(mayusculas, minusculas, numeros, signos):
    
    caracteres = ""
    obligatorios = []
    
    if mayusculas:
        caracteres += string.ascii_uppercase
        obligatorios.append(np.random.choice(list(string.ascii_uppercase)))
    if minusculas:
        caracteres += string.ascii_lowercase
        obligatorios.append(np.random.choice(list(string.ascii_lowercase)))
    if numeros:
        caracteres += string.digits
        obligatorios.append(np.random.choice(list(string.digits)))
    if signos:
        caracteres += string.punctuation
        obligatorios.append(np.random.choice(list(string.punctuation)))

    return caracteres, obligatorios

def mantener_consistencia(longitud, caracteres, obligatorios):

    if len(caracteres) == 0:
        return "Debes activar al menos un tipo de caracter"
    
    if longitud < len(obligatorios):
        return f"La longitud debe ser igual o mayor a la cantidad de tipos de caracteres activados ({longitud} caracteres)."
    
def crear_contrasenya(longitud, mayusculas=True, minusculas=True, numeros=True, signos=True):

    caracteres, obligatorios = definir_caracteres(mayusculas, minusculas, numeros, signos)
    error = mantener_consistencia(longitud, caracteres, obligatorios)

    if error is not None:
        return error
    
    while len(obligatorios) < longitud:
        obligatorios.append(np.random.choice(list(caracteres)))
                            
    np.random.shuffle(obligatorios)

    contrasenya = "".join(obligatorios)

    print(f"Tu contraseña es: {contrasenya}")

    pyperclip.copy(contrasenya)

    return contrasenya


```

```python
PATRON_RESULTADO = re.compile(r"^(\d+)\s*-\s*(\d+)$")

def leer_archivo(nombre_archivo):
    ruta = Path(nombre_archivo)

    if not ruta.exists():
        raise FileNotFoundError(f"El archivo {nombre_archivo} no fue encontrado.")
    
    with ruta.open('r', encoding='utf-8') as archivo:
        lineas = [linea.strip() for linea in archivo if linea.strip()]

    return lineas

def separar_linea_partido(linea):

    partes_tabulador = linea.split("\t")

    if len(partes_tabulador) == 3:
        equipo_local = partes_tabulador[0].strip()
        resultado = partes_tabulador[1].strip()
        equipo_visitante = partes_tabulador[2].strip()

    else:
        partes = linea.strip()

        indice_resultado = None

        for i, parte in enumerate(partes):
            if PATRON_RESULTADO.match(parte):
                indice_resultado = i
                break
            
        if indice_resultado is None:
            raise ValueError(f"No se ha podido encontrar un resultado en la línea {linea}.")
        
        equipo_local = " ".join(partes[:indice_resultado]).strip()
        resultado = partes[indice_resultado].strip()
        equipo_visitante = " ".join(partes[indice_resultado + 1:]).strip()

    if not equipo_local or not equipo_visitante:
        raise ValueError(f"En la línea {linea} no se ha encontrado ningun equipo")
    
    coincidencia = PATRON_RESULTADO.match(resultado)

    if coincidencia is None:
        raise ValueError(f"En la linea {linea} no se ha encontrado ningún resultado")
    
    goles_local = int(coincidencia.group(1))
    goles_visitante = int(coincidencia.group(2))

    return equipo_local, equipo_visitante, goles_local, goles_visitante

def crear_estadisticas():
    return {
        "partidos_ganados": 0,
        "partidos_empatados": 0,
        "partidos_perdidos": 0,
        "goles": 0,
        "goles_encajados": 0,
        "puntos": 0
    }

def actualizar_estadisticas(estadisticas, equipo, goles_marcados, goles_encajados):

    if equipo not in estadisticas:
        estadisticas[equipo] = crear_estadisticas()

    estadisticas[equipo]["goles"] += goles_marcados
    estadisticas[equipo]["goles_encajados"] += goles_encajados

    if goles_marcados > goles_encajados:
        estadisticas[equipo]["partidos_ganados"] += 1
        estadisticas[equipo]["puntos"] += 3

    elif goles_marcados == goles_encajados:
        estadisticas[equipo]["partidos_empatados"] += 1
        estadisticas[equipo]["puntos"] += 1

    else:
        estadisticas[equipo]["partidos_perdidos"] += 1


def procesar_partidos(lineas):
    estadisticas = {}
    lineas_invalidas = []

    for numero_linea, linea in enumerate(lineas, start=1):

        try:
            equipo_local, equipo_visitante, goles_local, goles_visitante = separar_linea_partido(linea)

        except ValueError as error:
            lineas_invalidas.append((numero_linea, linea, str(error)))

        actualizar_estadisticas(
            estadisticas,
            equipo_local,
            goles_local,
            goles_visitante
        )

        actualizar_estadisticas(
            estadisticas,
            equipo_visitante,
            goles_visitante,
            goles_local
        )

    return estadisticas, lineas_invalidas

def construir_clasificacion(estadisticas):

    df = pd.DataFrame.from_dict(
        estadisticas,
        orient='index'
    )

    df.index.name = "equipo"

    df["diferencia_goles"] = df["goles"] - df["goles_encajados"]

    df = df.sort_values(
        by=["puntos", "diferencia_goles", "goles"],
        ascending=[False, False, False]
    )

    return df

def extraer_resultados(df):
    goles_por_equipo = df["goles"].to_dict()
    equipo_mas_goleador = df["goles"].idxmax()
    equipo_mas_goleado = df["goles_encajados"].idxmax()

    return goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado

def mostrar_resultados(df, goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado, lineas_invalidas):
    print(f"CLASIFICACIÓN \n{df.to_string()}\n")
    print(f"\nGOLES POR EQUIPO \n{goles_por_equipo}\n")
    print(f"\nEQUIPO MÁS GOLEADOR: {equipo_mas_goleador}")
    print(f"EQUIPO MÁS GOLEADO: {equipo_mas_goleado}")

    if lineas_invalidas:
        print("LINEAS NO PROCESADAS")

        for numero_linea, linea, motivo in lineas_invalidas:
            print(f"Linea {numero_linea}: {linea} -> {motivo}")

def main(nombre_archivo):

    lineas = leer_archivo(nombre_archivo)

    estadisticas, lineas_invalidas = procesar_partidos(lineas)

    df = construir_clasificacion(estadisticas)

    goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado = extraer_resultados(df)

    mostrar_resultados(
        df,
        goles_por_equipo,
        equipo_mas_goleador,
        equipo_mas_goleado,
        lineas_invalidas
    )

    return df, goles_por_equipo, equipo_mas_goleador, equipo_mas_goleado, lineas_invalidas

```
