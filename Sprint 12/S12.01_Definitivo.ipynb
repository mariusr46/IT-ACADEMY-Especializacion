# Tasca S12.01 — APIs REST

En este notebook resuelvo los tres niveles del Sprint 12 utilizando  requests  y  pandas.


```python
import requests
import pandas as pd
```

# Nivel 1 — JSONPlaceholder

Primero consultamos los recursos  /posts ,  /users  y  /todos  mediante peticiones GET.


```python
base_url = "https://jsonplaceholder.typicode.com"

recursos = {
    "posts": "/posts",
    "users": "/users",
    "todos": "/todos"
}

for nombre, endpoint in recursos.items():
    response = requests.get(base_url + endpoint)
    datos = response.json()

    print(nombre)
    print("Cantidad:", len(datos))
    print("Código de estado:", response.status_code)
    print()
```

 # Error 404

Hacemos una petición a una publicación que no existe para comprovar el código 404.


```python
response = requests.get(base_url + "/posts/999999")

print("Código de estado:", response.status_code)
```

 # Petición POST

Creamos una nueva publicación ficticia.


```python
nuevo_post = {
    "title": "Mi nueva publicación",
    "body": "Este es el contenido de la publicación",
    "userId": 1
}

response = requests.post(
    base_url + "/posts",
    json=nuevo_post
)

print("Respuesta JSON:")
print(response.json())

print("Código de estado:", response.status_code)
```

 # Petición PATCH

Modificamos solamente el título de una publicación existente.


```python
cambio = {
    "title": "Título modificado"
}

response = requests.patch(
    base_url + "/posts/1",
    json=cambio
)

print("Respuesta JSON:")
print(response.json())

print("Código de estado:", response.status_code)
```

 # Petición DELETE

Eliminamos una publicación y mostramos la respueta de la API.


```python
response = requests.delete(base_url + "/posts/1")

print("Respuesta JSON:")
print(response.json())

print("Código de estado:", response.status_code)
```

# Nivel 2 — API pública real

He elegido la API de **TVmaze**, que permite consultar información sobre series de televisión.

Dos endpoints disponibles son:

-  /search/shows?q=batman : permite buscar series por nombre.
-  /schedule?country=ES&date=2026-06-15 : permite consultar la programación por país y fecha.

Algunos parámetros interesantes son:

-  q : texto que queremos buscar.
-  country : código del país.
-  date : fecha de la programación.

La API devuelve los resultados en formato JSON.


 # Petición GET

Buscamos series relacionadas con Batman y mostramos algunos campos.


```python
url = "https://api.tvmaze.com/search/shows"

params = {
    "q": "batman"
}

response = requests.get(url, params=params)

print("Código de estado:", response.status_code)
print("Content-Type:", response.headers.get("Content-Type"))
```

```python
datos = response.json()

for resultado in datos[:5]:
    serie = resultado["show"]

    print("Nombre:", serie["name"])
    print("Idioma:", serie["language"])
    print("Géneros:", serie["genres"])
    print()
```

 # Conversión a DataFrame

Extraemos la información de cada serie y la convertimos a un DataFrame.


```python
series = []

for resultado in datos:
    series.append(resultado["show"])

df_tvmaze = pd.DataFrame(series)

df_tvmaze.head()
```

# Nivel 3 — Open Data BCN

En este nivel utilizamos la API de Open Data BCN.

He elegido el dataset del **arbolado de zona de Barcelona**, ya que dispone de recursos CSV y permite consultar sus datos mediante la API.

Primero buscamos el dataset mediante  package_search .


```python
url_search = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search"

params = {
    "q": "arbrat zona"
}

response = requests.get(url_search, params=params)

print("Código de estado:", response.status_code)

datos_busqueda = response.json()
datasets = datos_busqueda["result"]["results"]

print("Datasets encontrados:", len(datasets))
```

```python
for dataset in datasets[:10]:
    print(dataset["title"])
```

 # package_show

De los resultados buscamos el dataset  arbrat-zona  y consultamos todos sus recursos.


```python
dataset_id = None

for dataset in datasets:
    if dataset["name"] == "arbrat-zona":
        dataset_id = dataset["name"]
        break

if dataset_id is None:
    raise ValueError("No se ha encontrado el dataset arbrat-zona.")

print("Dataset seleccionado:", dataset_id)
```

```python
url_show = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_show"

params = {
    "id": dataset_id
}

response = requests.get(url_show, params=params)

datos_dataset = response.json()
recursos = datos_dataset["result"]["resources"]

print("Código de estado:", response.status_code)
print("Dataset:", datos_dataset["result"]["title"])
print("Número de recursos:", len(recursos))
```

 # Recursos CSV o JSON

Mostramos los recursos disponibles en estos dos formatos.


```python
for recurso in recursos:
    if recurso["format"].lower() in ["csv", "json"]:
        print("Nombre:", recurso["name"])
        print("Formato:", recurso["format"])
        print("Resource ID:", recurso["id"])
        print("DataStore activo:", recurso.get("datastore_active"))
        print()
```

 # Selección del recurso

Para utilizar  datastore_search  necesitamos un recurso que tenga el DataStore activo.

Seleccionamos el primer recurso CSV que cumple esta condición.


```python
resource_id = None

for recurso in recursos:
    if recurso["format"].lower() == "csv" and recurso.get("datastore_active") == True:
        resource_id = recurso["id"]
        print("Recurso seleccionado:", recurso["name"])
        print("Resource ID:", resource_id)
        break

if resource_id is None:
    raise ValueError("No se ha encontrado ningún recurso CSV con DataStore activo.")
```

 # datastore_search

Con el  resource_id  seleccionado recuperamos 100 registros.

También comprobamos que la respuesta tenga  success = True  antes de intentar acceder a  result , así evitamos que el código falle sin saber el motivo.


```python
url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"

params = {
    "resource_id": resource_id,
    "limit": 100
}

response = requests.get(url_datastore, params=params)

datos = response.json()

print("Código de estado:", response.status_code)
print("Success:", datos.get("success"))
```

```python
if datos.get("success") == True and "result" in datos:
    registros = datos["result"]["records"]

    print("Registros recuperados:", len(registros))
    print("Registros totales disponibles:", datos["result"]["total"])
else:
    print(datos)
    raise ValueError("La consulta a datastore_search no ha funcionado correctamente.")
```

 # Conversión a DataFrame

Convertimos los 100 registros obtenidos en un DataFrame de pandas.


```python
df_bcn = pd.DataFrame(registros)

df_bcn.head()
```

```python
df_bcn.to_csv(
    "open_data_bcn_100_registros.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Archivo guardado correctamente.")
```
